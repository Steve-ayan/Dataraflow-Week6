````python
# ==========================================
# Task 1: Write a NumPy program to test element-wise for NaN of a given array.
# ==========================================

import numpy as np

# Create a sample array containing numbers and NaN values
arr_nan = np.array([10.5, np.nan, 20.1, np.nan, 30.0])

# Test element-wise for NaN using np.isnan()
nan_check = np.isnan(arr_nan)

print("Original Array:", arr_nan)
print("Element-wise NaN test result:", nan_check)
````

````python
# ==========================================
# Task 2: Write a NumPy program to test element-wise for complex number, 
# real number of a given array. Also test whether a given number is a scalar type or not.
# ==========================================

import numpy as np

# Create an array with real and complex numbers
arr_mixed = np.array([2.5, 1 + 2j, 3.0, 4 - 5j, 7])

# Test for real numbers
real_check = np.isreal(arr_mixed)

# Test for complex numbers
complex_check = np.iscomplex(arr_mixed)

# Test whether specific items are scalar types
scalar_test_1 = np.isscalar(arr_mixed[0])
scalar_test_2 = np.isscalar(5.5)

print("Original Array:", arr_mixed)
print("Is Real element-wise:", real_check)
print("Is Complex element-wise:", complex_check)
print("Is arr_mixed[0] a scalar type?", scalar_test_1)
print("Is 5.5 a scalar type?", scalar_test_2)
````

````python
# ==========================================
# Task 3: Write a NumPy program to test whether two arrays are element-wise equal within a tolerance.
# ==========================================

import numpy as np

arr1 = np.array([10.0, 20.001, 30.0])
arr2 = np.array([10.0, 20.002, 30.0])

# Test element-wise equality within a tolerance using np.isclose()
tolerance_check = np.isclose(arr1, arr2, rtol=1e-05, atol=1e-02)

# Test if all elements are close within tolerance using np.allclose()
all_close_check = np.allclose(arr1, arr2, rtol=1e-05, atol=1e-02)

print("Array 1:", arr1)
print("Array 2:", arr2)
print("Element-wise close test:", tolerance_check)
print("Are both arrays entirely close within tolerance?", all_close_check)
````

````python
# ==========================================
# Task 4: Write a NumPy program to create an element-wise comparison 
# (greater, greater_equal, less and less_equal) of two given arrays.
# ==========================================

import numpy as np

a = np.array([12, 42, 33, 41])
b = np.array([10, 42, 35, 40])

print("Array a:", a)
print("Array b:", b)

# Greater than
print("a > b:", np.greater(a, b))

# Greater than or equal to
print("a >= b:", np.greater_equal(a, b))

# Less than
print("a < b:", np.less(a, b))

# Less than or equal to
print("a <= b:", np.less_equal(a, b))
````

````python
# ==========================================
# Task 5: Write a NumPy program to create an element-wise comparison 
# (equal, equal within a tolerance) of two given arrays.
# ==========================================

import numpy as np

x = np.array([1.0, 2.0, 3.0])
y = np.array([1.0, 2.0001, 3.0])

# Strict element-wise equality
strict_equal = np.equal(x, y)

# Equality within tolerance
tolerance_equal = np.isclose(x, y)

print("Array x:", x)
print("Array y:", y)
print("Strictly equal (np.equal):", strict_equal)
print("Equal within tolerance (np.isclose):", tolerance_equal)
````

### SQL Task 2 · Filtering & NULLs
For **London** (`week06.london_2014`), return **one row** with three counts:
1. `no_event_days`: days with **no event recorded**
2. `rain_days`: days whose event mentions **Rain** anywhere (e.g. `Rain`, `Fog-Rain`, `Rain-Thunderstorm`)
3. `wet_but_no_event_days`: days with **no event recorded** but `Precipitationmm > 0`

⚠️ The events column name starts with a **space**: `[ Events]`. And remember, `= NULL` never matches.
*Revisit: §1.2, §3.2, §3.3.*

### SQL Task 3 · CASE & conditional aggregation
From `week16.customer_churn_prediction`, for each `Contract_Type` return: number of `customers`, number
`churned` (`Churn = 1`), and `churn_rate_pct` (percentage, **rounded to 1 decimal place**).

**Return:** `Contract_Type`, `customers`, `churned`, `churn_rate_pct`, sorted by `churn_rate_pct` (highest first).
⚠️ Watch out for the integer-division trap. *Revisit: §4.4, §5.2.*

````python
SQL_03 = """
SELECT 
    Contract_Type,
    COUNT(*) AS customers,
    SUM(CASE WHEN Churn = 1 THEN 1 ELSE 0 END) AS churned,
    ROUND(SUM(CASE WHEN Churn = 1 THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 1) AS churn_rate_pct
FROM 
    week16.customer_churn_prediction
GROUP BY 
    Contract_Type
ORDER BY 
    churn_rate_pct DESC
"""

# Register your answer for grading
answer('s03', SQL_03)
````

### SQL Task 5 · JOIN across two sources
Join WHO TB data (`week06.who_pop_tb_all`) with World Bank population (`week08.wb_pop_2013`) on the country
name. Compute **TB deaths per 100,000 people** using the **World Bank** population (`[SP.POP.TOTL]`),
**rounded to 1 decimal place**.

**Return:** the **top 5** countries: `Country`, `tb_deaths_per_100k`, highest first.
*Revisit: §7.2–7.3.*

````python
SQL_05 = """
SELECT TOP 5
    t.Country,
    ROUND(t."TB deaths" * 100000.0 / w."SP.POP.TOTL", 1) AS tb_deaths_per_100k
FROM 
    week06.who_pop_tb_all t
JOIN 
    week08.wb_pop_2013 w ON t.Country = w.country_name
WHERE 
    w."SP.POP.TOTL" > 0 AND t."TB deaths" IS NOT NULL
ORDER BY 
    ROUND(t."TB deaths" * 100000.0 / w."SP.POP.TOTL", 1) DESC
"""

# Register your answer for grading
answer('s05', SQL_05)
````

*Revisit: §7.4. Bonus thought: why do these countries fail to match?*

### SQL Task 7 · UNION ALL
Stack the **six** city weather tables in `week06` and compute each city's **average `Mean TemperatureC`**
for 2014, **rounded to 1 decimal place**. Use exactly these city labels:
`'Beijing'`, `'Brasilia'`, `'Cape Town'`, `'Delhi'`, `'London'`, `'Moscow'`.

**Return:** `city`, `avg_mean_temp_c`, sorted by `city` (A→Z).
⚠️ `Mean TemperatureC` is stored as a whole number, so watch the integer-division trap. *Revisit: §8.1–8.2.*

````python
SQL_07 = """
SELECT 'Beijing' AS city, ROUND(CAST(AVG("Mean TemperatureC") * 1.0 AS NUMERIC), 1) AS avg_mean_temp_c FROM week06.beijing_pek_2014
UNION ALL
SELECT 'Brasilia' AS city, ROUND(CAST(AVG("Mean TemperatureC") * 1.0 AS NUMERIC), 1) AS avg_mean_temp_c FROM week06.brasilia_bsb_2014
UNION ALL
SELECT 'Cape Town' AS city, ROUND(CAST(AVG("Mean TemperatureC") * 1.0 AS NUMERIC), 1) AS avg_mean_temp_c FROM week06.capetown_cpt_2014
UNION ALL
SELECT 'Delhi' AS city, ROUND(CAST(AVG("Mean TemperatureC") * 1.0 AS NUMERIC), 1) AS avg_mean_temp_c FROM week06.delhi_del_2014
UNION ALL
SELECT 'London' AS city, ROUND(CAST(AVG("Mean TemperatureC") * 1.0 AS NUMERIC), 1) AS avg_mean_temp_c FROM week06.london_2014
UNION ALL
SELECT 'Moscow' AS city, ROUND(CAST(AVG("Mean TemperatureC") * 1.0 AS NUMERIC), 1) AS avg_mean_temp_c FROM week06.moscow_svo_2014
ORDER BY city ASC
"""
````

### SQL Task 8 · CTE + window ranking (top-1 per group)
For **London**, find the **hottest day of each month** (highest `Max TemperatureC`). If two days tie,
take the **earlier** date.

**Return:** `month_no`, `day` (a real **date**, convert `GMT` with `TRY_CONVERT(date, GMT)`), `max_temp`,
sorted by `month_no`. *Revisit: §10, §11.3.*

````python
SQL_08 = """
WITH ranked_days AS (
    SELECT 
        EXTRACT(MONTH FROM CAST(GMT AS DATE)) AS month_no,
        CAST(GMT AS DATE) AS day,
        "Max TemperatureC" AS max_temp,
        ROW_NUMBER() OVER (
            PARTITION BY EXTRACT(MONTH FROM CAST(GMT AS DATE)) 
            ORDER BY "Max TemperatureC" DESC, GMT ASC
        ) as rn
    FROM 
        week06.london_2014
)
SELECT 
    month_no,
    day,
    max_temp
FROM 
    ranked_days
WHERE 
    rn = 1
ORDER BY 
    month_no ASC
"""
````

### SQL Task 9 · Running total
From `week09.comtrade_milk_uk_monthly_14`, compute the UK's **total milk exports per month** (exclude `'World'`)
and a **year-to-date running total**.

**Return:** `Period`, `exports_usd`, `ytd_exports_usd`, sorted by `Period`. *Revisit: §11.4–11.5.*

````python
SQL_09 = """
WITH monthly AS (
    SELECT 
        period AS Period,
        SUM(primary_value) AS exports_usd
    FROM 
        week09.comtrade_milk_uk_monthly_14
    WHERE 
        flow = 'Export'
        AND Partner != 'World'
    GROUP BY 
        period
)
SELECT 
    Period,
    exports_usd,
    SUM(exports_usd) OVER (ORDER BY Period ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW) AS ytd_exports_usd
FROM 
    monthly
ORDER BY 
    Period ASC
"""
````
